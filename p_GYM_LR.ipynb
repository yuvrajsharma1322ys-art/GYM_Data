```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('gymdata.csv', index_col = 'Unnamed: 0')
sns.relplot(data=df, x='id', y='pulse', kind='scatter')
y = df['pulse'] # target decided
x = df.drop(['pulse', 'id'], axis = 1)  # these 2 columns get drop because id is not useful and pulse column is already assigned to target so remaining columns get assigned to x except these 2
x = pd.get_dummies(x, drop_first=True) # this can convert remaining columns(diet, time, kind) into integer value(0,1)
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x , y , test_size = 0.2, random_state = 42 )
model = LinearRegression()
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
plt.scatter(y_test, y_pred, color='green')
plt.xlabel('Actual Pulse')
plt.ylabel('Predicted Pulse')
plt.title('Actual vs Predicted Pulse')
plt.show()
```
